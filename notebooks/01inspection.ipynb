```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv(
    r"C:\Users\Pc\OneDrive\Documents\atomcamp\Pakistan_Ecommerceproj\Pakistan Largest Ecommerce Dataset.csv",
    low_memory=False
)
print("Dataset loaded successfully!")
df.head()
```

```python
print("=== DATASET DIMENSIONS ===")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("=== COLUMN NAMES ===")

print(df.columns.tolist())
```

```python
print("=== DATASET INFORMATION ===")

df.info()
```

```python
print("=== FIRST FIVE ROWS ===")

display(df.head())

print("=== LAST FIVE ROWS ===")

display(df.tail())
```

```python
print("=== NUMERICAL SUMMARY ===")

display(df.describe())

print("=== COMPLETE DATA SUMMARY ===")

display(df.describe(include="all").T)
```

## Checing Missing Values

```python
missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(
    missing_report.sort_values(
        by="Missing Percentage",
        ascending=False
    )
)
```

## Completely empty rows and columns

```python
empty_rows = df.isnull().all(axis=1).sum()

empty_columns = df.columns[
    df.isnull().all()
].tolist()

print("Completely empty rows:", empty_rows)

print("Completely empty columns:", empty_columns)
```

## DUPLICATE ANALYSIS

```python
print("=== DUPLICATE ANALYSIS ===")

print(
    "Total duplicate rows:",
    df.duplicated().sum()
)

df_nonempty = df.dropna(how="all")

print(
    "Duplicate rows after excluding completely empty rows:",
    df_nonempty.duplicated().sum()
)
```

## Inspect categorical columns

```python
categorical_columns = df.select_dtypes(
    include=["object", "str"]
).columns

for column in categorical_columns:

    print(f"\n=== {column} ===")

    print("Unique values:", df[column].nunique())

    print(df[column].value_counts().head(15))
```

## Inspect numerical columns

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns

print("Numerical columns:")

print(numerical_columns.tolist())

display(
    df[numerical_columns].describe().T
)
```
