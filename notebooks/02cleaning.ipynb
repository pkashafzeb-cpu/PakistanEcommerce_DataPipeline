# Data Cleaning and Validation

## Objective
Prepare the raw e-commerce dataset for analysis by identifying
and handling empty rows, empty columns, duplicate records,
missing values, and inconsistent data.

The original dataset will remain unchanged.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    r"C:\Users\Pc\OneDrive\Documents\atomcamp\Pakistan_Ecommerceproj\Pakistan Largest Ecommerce Dataset.csv",
    low_memory=False
)
print("Original dataset shape:", df.shape)
```

## Remove completely empty rows and columns

```python
# Remove rows where every column is missing
df_clean = df.dropna(how="all")

# Remove columns where every value is missing
df_clean = df_clean.dropna(axis=1, how="all")

print("Shape after removing empty rows and columns:", df_clean.shape)
print("Remaining columns:", df_clean.columns.tolist())
```

```python
print("Remaining completely empty rows:",
      df_clean.isnull().all(axis=1).sum())

print("Remaining completely empty columns:",
      df_clean.columns[df_clean.isnull().all()].tolist())

print("Remaining duplicate rows:",
      df_clean.duplicated().sum())

print("New dataset shape:", df_clean.shape)
```

## Inspect Missing Values

```python
# Count missing values in each column
missing_values = df_clean.isnull().sum()

# Calculate missing percentages
missing_percentage = (
    df_clean.isnull().mean() * 100
)

# Create a report
missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

# Sort by missing percentage
missing_report = missing_report.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_report)
```

```python
# Inspect records with missing commission codes
display(
    df_clean[df_clean["sales_commission_code"].isnull()]
    .head(10)
)
```

```python
columns_to_check = [
    "category_name_1",
    "sku",
    "status",
    "Customer Since",
    "Customer ID"
]

for column in columns_to_check:
    print(f"\nMissing records in: {column}")

    display(
        df_clean[df_clean[column].isnull()]
        .head(5)
    )
```

```python
# Check special missing-value representations
for column in df_clean.select_dtypes(include="object").columns:
    special_values = df_clean[column].astype(str).str.strip().isin(
        ["\\N", "", "NULL", "null", "None"]
    ).sum()

    if special_values > 0:
        print(column, ":", special_values)
```

```python
#Inspect all 15 missing statuses
missing_status = df_clean[df_clean["status"].isnull()]

display(missing_status)
```

```python
#Inspect missing categories
missing_category = df_clean[
    df_clean["category_name_1"].isnull()
]

display(missing_category)
```

```python
#Check commission-code distribution
print(
    df_clean["sales_commission_code"]
    .value_counts(dropna=False)
    .head(20)
)
```

```python
# Find special missing-value markers in text columns

text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    count = (
        df_clean[column]
        .astype("string")
        .str.strip()
        .isin(["\\N", "", "NULL", "null"])
        .sum()
    )

    if count > 0:
        print(f"{column}: {count}")
```

```python
# Convert special missing-value markers into actual missing values

for column in text_columns:
    df_clean[column] = df_clean[column].astype("string").str.strip()

    df_clean[column] = df_clean[column].replace({
        "\\N": pd.NA,
        "": pd.NA,
        "NULL": pd.NA,
        "null": pd.NA
    })
```

```python
missing_report = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": df_clean.isnull().mean() * 100
}).sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_report)
```

```python
for column in text_columns:
    count = (
        df_clean[column]
        .astype("string")
        .str.strip()
        .isin(["\\N", "", "NULL", "null"])
        .sum()
    )

    if count > 0:
        print(f"{column}: {count}")
```

## Validation Before Data Type Conversion.

This means:
- MM/DD/YYYY successfully interpreted every date.
- DD/MM/YYYY failed on 422,034 records.
Therefore, we should use month/day/year for both columns.

```python
# Check whether date values can be parsed correctly

date_columns = ["created_at", "Working Date"]

for column in date_columns:
    print(f"\n{column}")

    us_dates = pd.to_datetime(
        df_clean[column],
        format="%m/%d/%Y",
        errors="coerce"
    )

    day_first_dates = pd.to_datetime(
        df_clean[column],
        format="%d/%m/%Y",
        errors="coerce"
    )

    print("Invalid with month/day/year:", us_dates.isna().sum())
    print("Invalid with day/month/year:", day_first_dates.isna().sum())


# Check Customer Since separately
customer_dates = pd.to_datetime(
    df_clean["Customer Since"],
    format="%Y-%m",
    errors="coerce"
)

print("\nCustomer Since")
print("Invalid dates:", customer_dates.isna().sum())
```

```python
# Check for decimal values in whole-number columns

numeric_columns = [
    "item_id",
    "qty_ordered",
    "Year",
    "Month",
    "Customer ID"
]

for column in numeric_columns:
    values = df_clean[column].dropna()

    decimal_count = (values % 1 != 0).sum()

    print(f"{column}: {decimal_count} decimal values")
```

## Data Type Conversion

```python
# Convert date columns into datetime format

df_clean["created_at"] = pd.to_datetime(
    df_clean["created_at"],
    format="%m/%d/%Y"
)

df_clean["Working Date"] = pd.to_datetime(
    df_clean["Working Date"],
    format="%m/%d/%Y"
)

# Convert Customer Since into monthly datetime values

df_clean["Customer Since"] = pd.to_datetime(
    df_clean["Customer Since"],
    format="%Y-%m"
)

# Convert whole-number columns into nullable integers

integer_columns = [
    "item_id",
    "qty_ordered",
    "Year",
    "Month",
    "Customer ID"
]

for column in integer_columns:
    df_clean[column] = df_clean[column].astype("Int64")

print("Data type conversion completed.")
```

## Categorical Data Inspection
We inspect categorical columns to identify unique values, inconsistent labels, and unexpected entries before standardizing them.

```python
# Identify unique values in categorical columns

categorical_columns = [
    "status",
    "category_name_1",
    "payment_method",
    "BI Status"
]

for column in categorical_columns:
    print(f"\n{column}")
    print("Unique values:", df_clean[column].nunique())
    print(df_clean[column].value_counts(dropna=False).head(20))
```

```python
# Find the record containing the suspicious BI Status

display(
    df_clean[df_clean["BI Status"] == "#REF!"]
)
```

```python
# Replace the invalid BI Status value

df_clean["BI Status"] = df_clean["BI Status"].replace(
    "#REF!", pd.NA
)

print("Remaining #REF! values:",
      (df_clean["BI Status"] == "#REF!").sum())

print("Missing BI Status values:",
      df_clean["BI Status"].isna().sum())
```

```python
print("Final shape:", df_clean.shape)
print("Duplicate rows:", df_clean.duplicated().sum())
print("Completely empty rows:", df_clean.isna().all(axis=1).sum())
print("Completely empty columns:", df_clean.isna().all(axis=0).sum())

print("\nMissing values:")
display(df_clean.isna().sum().sort_values(ascending=False))
```
